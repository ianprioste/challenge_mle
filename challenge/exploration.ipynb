## 2. Features Generation

### 2.a. Period of Day

```python
from datetime import datetime

def get_period_day(date):
    try:
        date_time = datetime.strptime(date, '%Y-%m-%d %H:%M:%S').time()
        if datetime.strptime("05:00", '%H:%M').time() <= date_time <= datetime.strptime("11:59", '%H:%M').time():
            return 'mañana'
        elif datetime.strptime("12:00", '%H:%M').time() <= date_time <= datetime.strptime("18:59", '%H:%M').time():
            return 'tarde'
        else:
            return 'noche'
        
    except ValueError:
        return 'Invalid date'
```

```python
data['period_day'] = data['Fecha-I'].apply(get_period_day)
```

### 2.b. High Season

```python
from datetime import datetime

def is_high_season(fecha):
    try:
        fecha = datetime.strptime(fecha, '%Y-%m-%d %H:%M:%S')
        ranges = [
            (datetime(fecha.year, 12, 15), datetime(fecha.year, 12, 31)),
            (datetime(fecha.year, 1, 1), datetime(fecha.year, 3, 3)),
            (datetime(fecha.year, 7, 15), datetime(fecha.year, 7, 31)),
            (datetime(fecha.year, 9, 11), datetime(fecha.year, 9, 30))
        ]
        return any(start <= fecha <= end for start, end in ranges)
    except ValueError:
        return False
```

```python
data['high_season'] = data['Fecha-I'].apply(is_high_season)
```

### 2.c. Difference in Minutes

```python
def get_min_diff(data: pd.DataFrame) -> pd.Series: 
    try:
        fecha_o = datetime.strptime(data['Fecha-O'], '%Y-%m-%d %H:%M:%S')
        fecha_i = datetime.strptime(data['Fecha-I'], '%Y-%m-%d %H:%M:%S')
        return (fecha_o - fecha_i).total_seconds() / 60
    except ValueError:
        return None
```

```python
data['min_diff'] = data.apply(get_min_diff, axis = 1)
```

### 2.d. Delay

```python
threshold_in_minutes = 15
data['delay'] = np.where(data['min_diff'] > threshold_in_minutes, 1, 0)
```

## 3. Data Analysis: Second Sight

### How is the delay rate across columns?

```python
def get_rate_from_column(data, column):
    delays = data[data['delay'] == 1][column].value_counts().to_dict()
    total = data[column].value_counts().to_dict()
    
    rates = {name: round(delays.get(name, 0) / total*100, 2) for name, total in total.items()}
    
    return pd.DataFrame.from_dict(data=rates, orient='index', columns=['Tasa (%)'])
```

```python
destination_rate = get_rate_from_column(data, 'SIGLADES')
destination_rate_values = data['SIGLADES'].value_counts().index
destination_rate = destination_rate.sort_values(by='Tasa (%)', ascending=True)


plt.figure(figsize = (15,20))
sns.set(style="darkgrid")
sns.barplot(y=destination_rate_values, x=destination_rate['Tasa (%)'], palette=palette, alpha = 0.75)
plt.title('Delay Rate by Destination')
plt.ylabel('Delay Rate [%]', fontsize=12)
plt.xlabel('Destination', fontsize=12)
plt.xticks(rotation=90)
plt.show()
```

## 4. Training

### 4.a. Data Split (Training and Validation)

```python
from sklearn.model_selection import train_test_split
from sklearn.utils import shuffle
from sklearn.metrics import confusion_matrix, classification_report
```

```python
training_data = shuffle(data[['OPERA', 'MES', 'TIPOVUELO', 'SIGLADES', 'DIANOM', 'delay']], random_state = 111)
```

```python
features = pd.concat([
    pd.get_dummies(data['OPERA'], prefix = 'OPERA'),
    pd.get_dummies(data['TIPOVUELO'], prefix = 'TIPOVUELO'), 
    pd.get_dummies(data['MES'], prefix = 'MES')], 
    axis = 1
)
target = data['delay']
```

```python
x_train, x_test, y_train, y_test = train_test_split(features, target, test_size = 0.33, random_state = 42)
```

```python
print(f"train shape: {x_train.shape} | test shape: {x_test.shape}")
```

```python
y_train.value_counts('%')*100
```

```python
y_test.value_counts('%')*100
```

### 4.b. Model Selection

#### 4.b.i. XGBoost

```python
import xgboost as xgb
from xgboost import plot_importance
```

```python
xgb_model = xgb.XGBClassifier(random_state=1, learning_rate=0.01)
xgb_model.fit(x_train, y_train)
```

```python
xgboost_y_preds = xgb_model.predict(x_test)
xgboost_y_preds = [1 if y_pred > 0.5 else 0 for y_pred in xgboost_y_preds]
```

```python
confusion_matrix(y_test, xgboost_y_preds)
```

```python
print(classification_report(y_test, xgboost_y_preds))
```

#### 4.b.ii. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
reg_model = LogisticRegression()
reg_model.fit(x_train, y_train)
```

```python
reg_y_preds = reg_model.predict(x_test)
```

```python
confusion_matrix(y_test, reg_y_preds)
```

```python
print(classification_report(y_test, reg_y_preds))
```

## 5. Data Analysis: Third Sight

### Feature Importance

```python
plt.figure(figsize = (10,5))
plot_importance(xgb_model)
```

```python
importances = xgb_model.feature_importances_

feature_importances = pd.DataFrame({'Feature': features.columns, 'Importance': importances})

top_10_features = feature_importances.sort_values(by='Importance', ascending=False).head(10)['Feature'].tolist()

print(top_10_features)
```

### Data Balance

```python
n_y0 = len(y_train[y_train == 0])
n_y1 = len(y_train[y_train == 1])
scale = n_y0/n_y1
print(scale)
```
